### Exercise 01: Variables And Datatypes

```python

width = 640
height = 480
image_name = "sample.png"
is_color = True

print(width, type(width))
print(height, type(height))
print(image_name, type(image_name))
print(is_color, type(is_color))

total_pixels = width * height
print("Total pixels:", total_pixels)
```

### Exercise 02: if, elif, else



```python

intensities = [-1, 0, 127, 128, 255, 256]

for intensity in intensities:
    if intensity < 0 or intensity > 255:
        label = "Invalid"
    elif intensity >= 128:
        label = "Bright pixel"
    else:
        label = "Dark pixel"

    print(intensity, ":", label)
```

### Exercise 03: For Loop

```python

pixels = [20, 130, 255]

for pixel in pixels:
    if pixel >= 128:
        print(pixel, "Bright pixel")
    else:
        print(pixel, "Dark pixel")
```

### Exercise 03: While Loop

```python

count = 0

while count < 3:
    print(count)
    count += 1
```

### Exercise: 04

```python

def is_bright(pixel, threshold=128):
    return pixel >= threshold

print(is_bright(127))
print(is_bright(128))
print(is_bright(200, 210))
```

### Exercise 05: Lists And Tuples

```python

pixels = [0, 128, 255]

print("First item:", pixels[0])

pixels[1] = 150
pixels.append(64)

print("Updated list:", pixels)
print("Length:", len(pixels))

size = (640, 480)
width, height = size

print("Width:", width)
print("Height:", height)

# A list is mutable, so its items can be changed.
# A tuple is immutable, so its items cannot be reassigned.
```

### Exercise 06: Dictionaries And Sets

```python

image = {"width": 640, "height": 480}
image["mode"] = "RGB"

print("Width:", image["width"])
print("Mode:", image["mode"])

labels = ["cat", "dog", "cat", "bird"]
unique_labels = set(labels)

print("Unique labels:", unique_labels)
print("Number of unique labels:", len(unique_labels))
```
